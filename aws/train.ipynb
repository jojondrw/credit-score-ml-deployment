```python
import os
import json
import joblib
import pandas as pd
import sklearn
from preprocessing import Preprocessing
from train_model import RandomForestModel

print(sklearn.__version__)

train_df = pd.read_csv("train/train.csv")
X_train = train_df.drop(columns=["Credit_Score"])
y_train = train_df["Credit_Score"]
print("train_df shape :", train_df.shape)
print("X_train shape  :", X_train.shape)

with open("artifacts/best_params.json") as f:
    best_params = json.load(f)
print("Best params:", best_params)

prep = Preprocessing()
X_train_t = prep.fit_transform(X_train)
print("X_train_t shape:", X_train_t.shape)

model = RandomForestModel()
model.model.set_params(**best_params)
model.train(X_train_t, y_train)

os.makedirs("model", exist_ok=True)
joblib.dump(prep, "model/preprocessing.pkl", compress=9)
joblib.dump(model.model, "model/best_model.pkl", compress=9)

print("Model + preprocessing tersimpan di folder model/")
```
